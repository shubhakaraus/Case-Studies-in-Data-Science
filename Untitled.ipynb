```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer

from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    confusion_matrix,
    classification_report,
    roc_curve,
    precision_recall_curve
)

import os

# 1. COPY DATASET

shop = shoppers.copy()

print("=" * 70)
print("ONLINE SHOPPERS PURCHASING INTENTION")
print("=" * 70)

print("\nDataset shape:")
print(shop.shape)

# 2. CHECK MISSING VALUES

print("\nMissing values:")
print(shop.isnull().sum())


# 3. CONVERT TARGET VARIABLE

# Revenue:
# False = customer did not purchase
# True  = customer purchased

shop["Revenue"] = (
    shop["Revenue"]
    .astype(str)
    .str.lower()
    .map({
        "true": 1,
        "false": 0
    })
)

# Removing any rows where Revenue could not be converted
shop = shop.dropna(subset=["Revenue"])

shop["Revenue"] = shop["Revenue"].astype(int)


# 4. DISPLAY TARGET DISTRIBUTION

print("\nTarget distribution:")
print(shop["Revenue"].value_counts())

print("\nTarget distribution (%):")
print(
    shop["Revenue"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

print("\nInterpretation:")
print("0 = No purchase")
print("1 = Purchase")


# 5. SEPARATE FEATURES AND TARGET

X = shop.drop(columns=["Revenue"])
y = shop["Revenue"]


# 6. IDENTIFY NUMERICAL AND CATEGORICAL FEATURES

numeric_features = X.select_dtypes(
    include=["int64", "float64", "int32", "float32"]
).columns.tolist()

categorical_features = X.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

print("\nNumerical features:")
print(numeric_features)

print("\nCategorical features:")
print(categorical_features)



# 7. PREPROCESSING

numeric_transformer = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "scaler",
            StandardScaler()
        )
    ]
)

categorical_transformer = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="most_frequent")
        ),
        (
            "onehot",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            )
        )
    ]
)

preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            numeric_transformer,
            numeric_features
        ),
        (
            "cat",
            categorical_transformer,
            categorical_features
        )
    ]
)


# 8. TRAIN / TEST SPLIT

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nTraining samples:", len(X_train))
print("Testing samples:", len(X_test))


# 9. RANDOM FOREST

print("\n" + "=" * 70)
print("TRAINING RANDOM FOREST")
print("=" * 70)

rf_model = Pipeline(
    steps=[
        (
            "preprocessor",
            preprocessor
        ),
        (
            "classifier",
            RandomForestClassifier(
                n_estimators=300,
                random_state=42,
                class_weight="balanced",
                n_jobs=-1
            )
        )
    ]
)

rf_model.fit(
    X_train,
    y_train
)

rf_predictions = rf_model.predict(
    X_test
)

rf_probabilities = rf_model.predict_proba(
    X_test
)[:, 1]

print("Random Forest training completed!")

# 10. SVM

print("\n" + "=" * 70)
print("TRAINING SVM")
print("=" * 70)

svm_model = Pipeline(
    steps=[
        (
            "preprocessor",
            preprocessor
        ),
        (
            "classifier",
            SVC(
                kernel="rbf",
                probability=True,
                class_weight="balanced",
                random_state=42
            )
        )
    ]
)

svm_model.fit(
    X_train,
    y_train
)

svm_predictions = svm_model.predict(
    X_test
)

svm_probabilities = svm_model.predict_proba(
    X_test
)[:, 1]

print("SVM training completed!")

# 11. EVALUATION FUNCTION

def calculate_metrics(
    model_name,
    y_true,
    predictions,
    probabilities
):

    return {
        "Model": model_name,

        "Accuracy": accuracy_score(
            y_true,
            predictions
        ),

        "Precision": precision_score(
            y_true,
            predictions,
            zero_division=0
        ),

        "Recall": recall_score(
            y_true,
            predictions,
            zero_division=0
        ),

        "F1_Score": f1_score(
            y_true,
            predictions,
            zero_division=0
        ),

        "ROC_AUC": roc_auc_score(
            y_true,
            probabilities
        ),

        "PR_AUC": average_precision_score(
            y_true,
            probabilities
        )
    }

# 12. CALCULATE MODEL METRICS

rf_metrics = calculate_metrics(
    "Random Forest",
    y_test,
    rf_predictions,
    rf_probabilities
)

svm_metrics = calculate_metrics(
    "SVM",
    y_test,
    svm_predictions,
    svm_probabilities
)

results_shop = pd.DataFrame(
    [rf_metrics, svm_metrics]
)

# 13. DISPLAY RESULTS

print("\n" + "=" * 70)
print("ONLINE SHOPPERS - MODEL PERFORMANCE")
print("=" * 70)

print(
    results_shop.to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)

# 14. SAVE RESULTS

os.makedirs(
    "part1_3_results",
    exist_ok=True
)

results_shop.to_csv(
    "part1_3_results/online_shoppers_model_results.csv",
    index=False
)

# 15. RANDOM FOREST CLASSIFICATION REPORT

print("\n" + "=" * 70)
print("RANDOM FOREST - CLASSIFICATION REPORT")
print("=" * 70)

print(
    classification_report(
        y_test,
        rf_predictions,
        target_names=[
            "No Purchase",
            "Purchase"
        ],
        digits=4
    )
)

# 16. SVM CLASSIFICATION REPORT

print("\n" + "=" * 70)
print("SVM - CLASSIFICATION REPORT")
print("=" * 70)

print(
    classification_report(
        y_test,
        svm_predictions,
        target_names=[
            "No Purchase",
            "Purchase"
        ],
        digits=4
    )
)

# 17. CONFUSION MATRICES

fig, axes = plt.subplots(
    1,
    2,
    figsize=(13, 5)
)

cm_rf = confusion_matrix(
    y_test,
    rf_predictions
)

cm_svm = confusion_matrix(
    y_test,
    svm_predictions
)

sns.heatmap(
    cm_rf,
    annot=True,
    fmt="d",
    cmap="Blues",
    ax=axes[0]
)

axes[0].set_title(
    "Random Forest - Online Shoppers"
)

axes[0].set_xlabel("Predicted")
axes[0].set_ylabel("Actual")

sns.heatmap(
    cm_svm,
    annot=True,
    fmt="d",
    cmap="Greens",
    ax=axes[1]
)

axes[1].set_title(
    "SVM - Online Shoppers"
)

axes[1].set_xlabel("Predicted")
axes[1].set_ylabel("Actual")

plt.tight_layout()

plt.savefig(
    "part1_3_results/online_shoppers_confusion_matrices.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# 18. ROC CURVE

fpr_rf, tpr_rf, _ = roc_curve(
    y_test,
    rf_probabilities
)

fpr_svm, tpr_svm, _ = roc_curve(
    y_test,
    svm_probabilities
)

plt.figure(figsize=(8, 6))

plt.plot(
    fpr_rf,
    tpr_rf,
    label=(
        f"Random Forest "
        f"(AUC = {roc_auc_score(y_test, rf_probabilities):.3f})"
    )
)

plt.plot(
    fpr_svm,
    tpr_svm,
    label=(
        f"SVM "
        f"(AUC = {roc_auc_score(y_test, svm_probabilities):.3f})"
    )
)

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    label="Random Classifier"
)

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")

plt.title(
    "Online Shoppers - ROC Curve"
)

plt.legend()
plt.grid(alpha=0.3)

plt.savefig(
    "part1_3_results/online_shoppers_ROC_curve.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# 19. PRECISION-RECALL CURVE

precision_rf, recall_rf, _ = precision_recall_curve(
    y_test,
    rf_probabilities
)

precision_svm, recall_svm, _ = precision_recall_curve(
    y_test,
    svm_probabilities
)

plt.figure(figsize=(8, 6))

plt.plot(
    recall_rf,
    precision_rf,
    label=(
        f"Random Forest "
        f"(PR-AUC = "
        f"{average_precision_score(y_test, rf_probabilities):.3f})"
    )
)

plt.plot(
    recall_svm,
    precision_svm,
    label=(
        f"SVM "
        f"(PR-AUC = "
        f"{average_precision_score(y_test, svm_probabilities):.3f})"
    )
)

plt.xlabel("Recall")
plt.ylabel("Precision")

plt.title(
    "Online Shoppers - Precision-Recall Curve"
)

plt.legend()
plt.grid(alpha=0.3)

plt.savefig(
    "part1_3_results/online_shoppers_precision_recall_curve.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# 20. RANDOM FOREST FEATURE IMPORTANCE

rf_preprocessor = rf_model.named_steps[
    "preprocessor"
]

rf_classifier = rf_model.named_steps[
    "classifier"
]

feature_names = (
    rf_preprocessor
    .get_feature_names_out()
)

feature_importance = pd.DataFrame({
    "Feature": feature_names,
    "Importance": rf_classifier.feature_importances_
})

feature_importance = (
    feature_importance
    .sort_values(
        "Importance",
        ascending=False
    )
)

# Make names easier to read
feature_importance["Feature"] = (
    feature_importance["Feature"]
    .str.replace(
        "num__",
        "",
        regex=False
    )
    .str.replace(
        "cat__",
        "",
        regex=False
    )
)

print("\n" + "=" * 70)
print("TOP 20 RANDOM FOREST FEATURES")
print("=" * 70)

print(
    feature_importance.head(20).to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)

feature_importance.to_csv(
    "part1_3_results/online_shoppers_feature_importance.csv",
    index=False
)

# 21. FEATURE IMPORTANCE GRAPH

top_features = (
    feature_importance
    .head(15)
    .sort_values("Importance")
)

plt.figure(figsize=(10, 7))

plt.barh(
    top_features["Feature"],
    top_features["Importance"]
)

plt.xlabel("Feature Importance")
plt.ylabel("Feature")

plt.title(
    "Online Shoppers - Top 15 Predictive Features"
)

plt.tight_layout()

plt.savefig(
    "part1_3_results/online_shoppers_feature_importance.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# 22. FINAL SUMMARY

print("\n" + "=" * 70)
print("ONLINE SHOPPERS ANALYSIS COMPLETED")
print("=" * 70)

print("\nResults saved in:")
print(
    os.path.abspath("part1_3_results")
)

print("\nFiles created:")
print("- online_shoppers_model_results.csv")
print("- online_shoppers_feature_importance.csv")
print("- online_shoppers_confusion_matrices.png")
print("- online_shoppers_ROC_curve.png")
print("- online_shoppers_precision_recall_curve.png")
print("- online_shoppers_feature_importance.png")
```
